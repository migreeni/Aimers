### pdf export example

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages

x = np.linspace(0, 2*np.pi, 400)

with PdfPages("plots_bundle.pdf") as pdf:
    # fig 1
    fig1 = plt.figure()
    plt.plot(x, np.sin(x))
    plt.title("sin")
    pdf.savefig(fig1)   # 현재 figure를 PDF에 추가
    plt.close(fig1)

    # fig 2
    fig2 = plt.figure()
    plt.plot(x, np.cos(x))
    plt.title("cos")
    pdf.savefig(fig2)
    plt.close(fig2)

    # fig 3 (subplots 예시)
    fig3, ax = plt.subplots(2, 2, figsize=(6,4))
    ax[0,0].plot(x, np.sin(x)); ax[0,0].set_title("sin")
    ax[0,1].plot(x, np.cos(x)); ax[0,1].set_title("cos")
    ax[1,0].plot(x, np.tan(x)); ax[1,0].set_ylim(-5,5); ax[1,0].set_title("tan")
    ax[1,1].plot(x, np.sinh(x)); ax[1,1].set_title("sinh")
    fig3.tight_layout()
    pdf.savefig(fig3)
    plt.close(fig3)
```
